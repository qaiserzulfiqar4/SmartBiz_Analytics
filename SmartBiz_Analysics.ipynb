```python
import pandas as pd
data = {
    "product":["Shirt","shoes","bags","watch","perfumes"],
    "Quantity": [10,5,8,12,4],
    "Price":[15000,4000,2500,1500,3000]
}
df=pd.DataFrame(data)
df["Revenue"] = df["Quantity"] * df["Price"]
df
```

```python
df["Cost_Per_Unit"] = [900, 2800, 1400, 900, 1800]

df["Total_Cost"] = df["Quantity"] * df["Cost_Per_Unit"]

df["Profit"] = df["Revenue"] - df["Total_Cost"]

print("Total Revenue:", df["Revenue"].sum())
print("Total Cost:", df["Total_Cost"].sum())
print("Total Profit:", df["Profit"].sum())

df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.bar(df["product"], df["Revenue"])
plt.title("SmartBiz Analytics - Product Revenue")
plt.xlabel("Product")
plt.ylabel("Revenue (Rs.)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
print("===== SMARTBIZ ANALYTICS =====")

print("\nDataset Overview:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
print(df.describe())
```

```python
product_sales = df.groupby("product")["Quantity"].sum()

best_product = product_sales.idxmax()
best_quantity = product_sales.max()

print("Best-Selling Product:", best_product)
print("Total Quantity Sold:", best_quantity)
```

```python
import matplotlib.pyplot as plt
Product_profit = df.groupby("product")["Profit"].sum()

Product_profit.plot(kind="bar", figsize=(8, 5))

plt.title("Product-wise Profit Analysis")
plt.xlabel("Product")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
comparison = df.groupby("product")[["Revenue", "Profit"]].sum()
comparison.plot(kind="bar", figsize=(9, 5))
plt.title("Revenue vs Profit Comparison")
plt.xlabel("product")
plt.ylabel("Amount")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
summary = df.groupby("product").agg(
    Total_Quantity=("Quantity", "sum"),
    Total_Revenue=("Revenue", "sum"),
    Total_Profit=("Profit", "sum")
).reset_index()

print(summary)
```
